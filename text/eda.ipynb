(1) shows the dataframe, (2) prints the number of rows in it, (3) prints the class labels in alphabetical order, (4) prints how many texts does each class has, (5) prints how many text and labels are null

```python
print(df)
print("number of rows ", len(df))
print("Class labels:", sorted(df["label"].unique()))
print(df["label"].value_counts())
print(df.isna().sum())
```

(3) puts the count of each class in label, (4) a list of colors, (5) creating the pie chart and giving it the counts, indexes of labels, percentages, number format and colors, (6) shows the pie

```python
import matplotlib.pyplot as plt
from collections import Counter

label = df["label"].value_counts()
colors = ["#82cc82", "#a897d5", "#dbb671", "#f3a79b", "#86cacd"]
plt.pie(label, labels=label.index, autopct="%1.1f%%", colors=colors)
plt.show()
```
